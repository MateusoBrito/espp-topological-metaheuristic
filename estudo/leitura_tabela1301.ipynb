# Leitura e Análise da Tabela 1301 do IBGE (`tabela1301.xlsx`)

Este notebook lê o arquivo `tabela1301.xlsx`, extraído do **SIDRA (Sistema IBGE de Recuperação Automática)**,
que contém dados do **Censo Demográfico do IBGE** referentes à área territorial.

```python
import pandas as pd

caminho_arquivo = "tabela1301.xlsx"

# 1. Inspecionar as abas (planilhas) disponíveis
excel_file = pd.ExcelFile(caminho_arquivo)
print("Abas encontradas no arquivo:", excel_file.sheet_names)
```

## 1. Visualização da Aba Principal (`Tabela`)
Vamos carregar os dados brutos e depois estruturados.

```python
# Leitura bruta da aba Tabela
df_tabela = pd.read_excel(caminho_arquivo, sheet_name="Tabela", header=None)
print("=== CONTEÚDO BRUTO DA PLANILHA ===")
display(df_tabela)
```

```python
# Extração dos metadados e valores limpos
titulo = df_tabela.iloc[0, 0]
variavel = df_tabela.iloc[1, 0]
ano = df_tabela.iloc[3, 1]
localidade = df_tabela.iloc[4, 0]
valor_area_km2 = df_tabela.iloc[4, 1]
fonte = df_tabela.iloc[5, 0]

print(f"Título: {titulo}")
print(f"Variável: {variavel}")
print(f"Ano de Referência: {ano}")
print(f"{localidade}: {valor_area_km2:,.1f} km²")
print(f"Fonte: {fonte}")
```

## 2. Visualização das Notas Metodológicas e Legendas (`Notas`)

```python
df_notas = pd.read_excel(caminho_arquivo, sheet_name="Notas", header=None)
df_notas.columns = ["Item / Símbolo", "Descrição"]
df_notas.fillna("", inplace=True)
display(df_notas)
```

### Observação sobre este arquivo:
Este arquivo contém apenas a **área territorial total do Brasil (8.502.728,3 km²)** referente ao Censo 2010.

Se você pretende cruzar a área/densidade com os municípios das antenas (ex: Divinópolis, Belo Horizonte, etc.), você pode exportar a Tabela 1301 do site do SIDRA/IBGE selecionando o nível territorial **"Municípios"** em vez de "Brasil".
